```python
import requests
import pandas as pd
```

```python
url = "https://archive-api.open-meteo.com/v1/archive"

ANO_INICIAL = "2015-01-01"
ANO_FINAL = "2025-12-31"

params = {
    "latitude": -12.54,
    "longitude": -55.72,
    "start_date": ANO_INICIAL,
    "end_date": ANO_FINAL,
    "hourly": [
        "temperature_2m",
        "precipitation"
    ]
}
```

```python
resposta = requests.get(url, params=params)

print(resposta.status_code)
```

```python
# Converte a resposta da API para um dicionário do Python
dados = resposta.json()
```

```python
# Cria um DataFrame (tabela) com os dados da API

df_clima = pd.DataFrame({

    "data_hora": dados["hourly"]["time"],

    "temperatura": dados["hourly"]["temperature_2m"],

    "precipitacao": dados["hourly"]["precipitation"]

})
```

```python
# Tabela com as regiões do Brasil e suas coordenadas de referência

regioes = pd.DataFrame({

    "regiao": [
        "Norte",
        "Nordeste",
        "Centro-Oeste",
        "Sudeste",
        "Sul"
    ],

    "latitude": [
        -3.119,
        -12.977,
        -15.793,
        -23.550,
        -25.428
    ],

    "longitude": [
        -60.021,
        -38.501,
        -47.882,
        -46.633,
        -49.273
    ]

})

regioes
```

```python
# Lista onde serão guardados os dados de todas as regiões

dados_clima = []
```

```python
for _, linha in regioes.iterrows():

    params = {
        "latitude": linha["latitude"],
        "longitude": linha["longitude"],
        "start_date": ANO_INICIAL,
        "end_date": ANO_FINAL,
        "hourly": [
            "temperature_2m",
            "precipitation"
        ]
    }

    resposta = requests.get(url, params=params)

    dados = resposta.json()

    df = pd.DataFrame({
        "data_hora": dados["hourly"]["time"],
        "temperatura": dados["hourly"]["temperature_2m"],
        "precipitacao": dados["hourly"]["precipitation"]
    })

    df["regiao"] = linha["regiao"]

    dados_clima.append(df)

    print(f'{linha["regiao"]} ✔')
```

```python
df_clima = pd.concat(dados_clima, ignore_index=True)

df_clima.head()
```

```python
# ==========================================================
# ETAPA 10 - ENGENHARIA DE ATRIBUTOS
# ==========================================================

# -----------------------------
# Chuva acumulada por dia
# -----------------------------

chuva_diaria = (
    df_clima
    .groupby(["regiao","ano","mes","dia"])["precipitacao"]
    .sum()
    .reset_index(name="chuva_dia")
)

# -----------------------------
# Temperatura média diária
# -----------------------------

temperatura_diaria = (
    df_clima
    .groupby(["regiao","ano","mes","dia"])["temperatura"]
    .mean()
    .reset_index(name="temperatura_media_dia")
)

# -----------------------------
# Chuva acumulada por mês
# -----------------------------

chuva_mensal = (
    df_clima
    .groupby(["regiao","ano","mes"])["precipitacao"]
    .sum()
    .reset_index(name="chuva_mes")
)

# -----------------------------
# Temperatura média por mês
# -----------------------------

temperatura_mensal = (
    df_clima
    .groupby(["regiao","ano","mes"])["temperatura"]
    .mean()
    .reset_index(name="temperatura_media_mes")
)
```

```python
# ==========================================================
# CLASSIFICAÇÃO DA TEMPERATURA
# ==========================================================

def classificar_temperatura(valor):

    if valor < 18:
        return "Frio"

    elif valor < 25:
        return "Ameno"

    elif valor < 30:
        return "Quente"

    else:
        return "Muito Quente"


df_clima["classe_temperatura"] = (
    df_clima["temperatura"]
    .apply(classificar_temperatura)
)
```

```python
# ==========================================================
# CLASSIFICAÇÃO DA PRECIPITAÇÃO
# ==========================================================

def classificar_chuva(valor):

    if valor == 0:
        return "Sem chuva"

    elif valor < 2.5:
        return "Fraca"

    elif valor < 10:
        return "Moderada"

    elif valor < 30:
        return "Forte"

    else:
        return "Muito Forte"


df_clima["classe_chuva"] = (
    df_clima["precipitacao"]
    .apply(classificar_chuva)
)
```

```python
# ==========================================================
# ESTAÇÃO DO ANO
# ==========================================================

def estacao(mes):

    if mes in [12,1,2]:
        return "Verão"

    elif mes in [3,4,5]:
        return "Outono"

    elif mes in [6,7,8]:
        return "Inverno"

    else:
        return "Primavera"


df_clima["estacao"] = (
    df_clima["mes"]
    .apply(estacao)
)
```

```python
# ==========================================================
# DIAS SEM CHUVA
# ==========================================================

df_clima["sem_chuva"] = (
    df_clima["precipitacao"] == 0
)
```

```python
# ==========================================================
# ALERTA DE CALOR
# ==========================================================

df_clima["onda_calor"] = (
    df_clima["temperatura"] >= 35
)
```

```python
# ==========================================================
# EVENTO DE CHUVA EXTREMA
# ==========================================================

df_clima["chuva_extrema"] = (
    df_clima["precipitacao"] >= 30
)
```

```python
# ==========================================================
# DIAS CONSECUTIVOS SEM CHUVA
# ==========================================================

import numpy as np

# Considera que não choveu quando a precipitação é igual a zero
df_clima["sem_chuva"] = df_clima["precipitacao"] == 0

# Ordena por região e data
df_clima = df_clima.sort_values(["regiao", "data_hora"]).reset_index(drop=True)

# Cria coluna
df_clima["horas_sem_chuva"] = 0

# Calcula sequência de horas sem chuva para cada região
for regiao in df_clima["regiao"].unique():

    mascara = df_clima["regiao"] == regiao

    contador = 0

    valores = []

    for chuva in df_clima.loc[mascara, "sem_chuva"]:

        if chuva:
            contador += 1
        else:
            contador = 0

        valores.append(contador)

    df_clima.loc[mascara, "horas_sem_chuva"] = valores

# Converte horas em dias
df_clima["dias_sem_chuva"] = df_clima["horas_sem_chuva"] / 24
```

```python
# ==========================================================
# EVENTOS EXTREMOS
# ==========================================================

# Chuva extrema
df_clima["evento_chuva_extrema"] = (
    df_clima["precipitacao"] >= 30
)

# Calor extremo
df_clima["evento_calor_extremo"] = (
    df_clima["temperatura"] >= 35
)

# Evento extremo (qualquer um)
df_clima["evento_extremo"] = (
    df_clima["evento_chuva_extrema"] |
    df_clima["evento_calor_extremo"]
)
```

```python
# ==========================================================
# ANOMALIA DE TEMPERATURA
# ==========================================================

# Média histórica da região
media_regiao = (
    df_clima
    .groupby("regiao")["temperatura"]
    .transform("mean")
)

# Diferença em relação à média
df_clima["anomalia_temperatura"] = (
    df_clima["temperatura"] - media_regiao
)
```

```python
# ==========================================================
# ÍNDICE DE RISCO CLIMÁTICO
# ==========================================================

df_clima["risco_climatico"] = (

      df_clima["dias_sem_chuva"] * 0.40

    + df_clima["anomalia_temperatura"].abs() * 0.30

    + df_clima["evento_extremo"].astype(int) * 30

)
```

```python
# ==========================================================
# APTIDÃO AGRÍCOLA
# ==========================================================

def calcular_aptidao(linha):

    temperatura = linha["temperatura"]

    chuva = linha["precipitacao"]

    score = 100

    if temperatura < 18:
        score -= 20

    elif temperatura > 34:
        score -= 20

    if chuva == 0:
        score -= 15

    elif chuva > 25:
        score -= 10

    return max(score, 0)

df_clima["aptidao_agricola"] = (
    df_clima.apply(calcular_aptidao, axis=1)
)
```

```python
# ==========================================================
# APTIDÃO AGRÍCOLA
# ==========================================================

def calcular_aptidao(linha):

    temperatura = linha["temperatura"]

    chuva = linha["precipitacao"]

    score = 100

    if temperatura < 18:
        score -= 20

    elif temperatura > 34:
        score -= 20

    if chuva == 0:
        score -= 15

    elif chuva > 25:
        score -= 10

    return max(score, 0)

df_clima["aptidao_agricola"] = (
    df_clima.apply(calcular_aptidao, axis=1)
)
```

```python
# ==========================================================
# SCORE ESG CLIMÁTICO
# ==========================================================

def score_esg(linha):

    score = 100

    score -= linha["dias_sem_chuva"] * 0.8

    score -= abs(linha["anomalia_temperatura"]) * 2

    if linha["evento_extremo"]:
        score -= 20

    return max(0, min(score,100))

df_clima["score_esg"] = (
    df_clima.apply(score_esg, axis=1)
)
```

```python
# ==========================================================
# RANKING CLIMÁTICO
# ==========================================================

ranking = (

    df_clima

    .groupby("regiao")

    .agg(

        Score_ESG=("score_esg","mean"),

        Aptidao=("aptidao_agricola","mean"),

        Risco=("risco_climatico","mean"),

        Temperatura=("temperatura","mean"),

        Chuva=("precipitacao","sum")

    )

    .sort_values("Score_ESG",ascending=False)

)

ranking
```

```python
# ==========================================================
# DASHBOARD EXECUTIVO - RESUMO POR REGIÃO
# ==========================================================

resumo_clima = (

    df_clima

    .groupby("regiao")

    .agg(

        # -------------------------------
        # TEMPERATURA
        # -------------------------------

        temperatura_media=("temperatura", "mean"),

        temperatura_maxima=("temperatura", "max"),

        temperatura_minima=("temperatura", "min"),

        # -------------------------------
        # PRECIPITAÇÃO
        # -------------------------------

        chuva_total=("precipitacao", "sum"),

        chuva_media=("precipitacao", "mean"),

        chuva_maxima=("precipitacao", "max"),

        # -------------------------------
        # EVENTOS
        # -------------------------------

        horas_sem_chuva=("sem_chuva", "sum"),

        dias_sem_chuva=("dias_sem_chuva", "max"),

        horas_onda_calor=("onda_calor", "sum"),

        eventos_chuva_extrema=("evento_chuva_extrema", "sum"),

        eventos_calor_extremo=("evento_calor_extremo", "sum"),

        eventos_extremos=("evento_extremo", "sum"),

        # -------------------------------
        # ANÁLISE CLIMÁTICA
        # -------------------------------

        anomalia_media=("anomalia_temperatura", "mean"),

        risco_climatico=("risco_climatico", "mean"),

        aptidao_agricola=("aptidao_agricola", "mean"),

        score_esg=("score_esg", "mean")

    )

    .round(2)

)

resumo_clima
```
